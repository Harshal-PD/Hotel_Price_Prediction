```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Assuming you already loaded your data like this:
df = pd.read_csv('data/raw/hotel_bookings.csv')

# If your target (price) column is not named 'price', replace it correctly!

# First check what columns are available
print(df.columns)

# Plotting price vs. number of guests
plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x='adults', y='adr', hue='children', size='babies', palette='coolwarm', sizes=(20, 200))
plt.title('Room Price (ADR) vs Number of Adults (and Children)')
plt.xlabel('Number of Adults')
plt.ylabel('Price per Night (ADR)')
plt.grid(True)
plt.show()
```
